```python
import pandas as pd

output  = "kanji;kana;english;class\n" #names of the columns
with open("japanese_core_2000_anki.txt") as f:
    lines = f.readlines()
    for line in lines:
        output+= ";".join(line.split("\t")[1:5]) + "\n" 
with open("anki_fixed.txt","w") as f:
    f.write(output)
df = pd.read_csv("anki_fixed.txt",sep=";")
df["english"] = df["english"].map(lambda x: x.split(","))
df = df.explode("english")
df.to_csv("wordlist.txt",index=False,sep=";")
```
